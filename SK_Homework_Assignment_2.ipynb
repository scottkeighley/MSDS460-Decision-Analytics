#### 1Bii: What is the meaning of the surplus variable value 22.5 in the 1st dual constraint with respect to the primal problem?

The 1st dual constraint corresponds to primal variable $x_1$ (product 1): 
$$8\pi_1 + 4\pi_2 + 2\pi_3 \ge 25$$

At the optimal dual solution, this constraint has a surplus of 22.5: 
$$8(3.125) + 4(5.625) + 2(0) = 25 + 22.5 + 0 = 47.5$$

- LHS = 47.5 (opportunity cost of producing one unit of Product 1)
- RHS = 25 (revenue generated by one unit of Product 1)
- surplus = 47.5 - 25 = 22.5

Each $\pi_i$ is the shadow price of resource $i$. The coefficients 8, 4, 2 are the amounts of Resources 1, 2, 3 that one unit of Product 1 consumes. So the left-hand side is the total implicit (opportunity) cost of producing one unit of Product 1, valued at resource shadow prices. 

Producing one unit of Product 1 would consume resources whose combined shadow price value is 47.50 but only generate 25.00 in revenue. This is a net loss of 22.50 per unit produced. So the surplus of 22.5 is telling us that Product 1 is 22.50 too expensive to produce relative to what it earns.

```python
import networkx as nx

G = nx.DiGraph()

edges = [
    (1, 2, 7000), (1, 3, 15000), (1, 4, 21000), (1, 5, 27000), (1, 6, 35000),
    (2, 3, 7000), (2, 4, 15000), (2, 5, 21000), (2, 6, 27000),
    (3, 4, 7000), (3, 5, 15000), (3, 6, 21000),
    (4, 5, 7000), (4, 6, 15000),
    (5, 6, 7000)
]

for u, v, cost in edges: 
    G.add_edge(u, v, weight=cost)

path = nx.shortest_path(G, source=1, target=6, weight='weight')
cost = nx.shortest_path_length(G, source=1, target=6, weight='weight')

print(f"Optimal Path: {path}")
print(f"Minimum Cost: ${cost:,}")
```

The program director of a local community college needs help planning course offerings for next year. It has three professors who each teach four courses over the year, and four sections of calculus, AI/ML, and simulation must be offered next year. At least one section of each class must be offered during each semester (fall and spring). The professors have preferences regarding the semester and the course offerings (shown below). 

A professor's total satisfaction for teaching a class is the sum of the semester satisfaction and the course satisfaction. So professor 1 has a total satisfaction of 1 + 1 = 2 from teaching calculus during the fall semester. 

| | Professor 1 | Professor 2 | Professor 3 |
| --- | --- | --- | --- | 
| Fall Term | 1 | 3 | 1 |
| Spring Term | 2 | 1 | 2 |

| | Professor 1 | Professor 2 | Professor 3 | 
| --- | --- | --- | --- |
| Calculus | 1 | 3 | 3 |
| AI/ML | 5 | 2 | 5 |
| Simulation | 3 | 5 | 4 |

Part A: Formulate the problem as a minimum cost network flow problem that can be used to assign professors to courses to maximize the total satisfaction of the three professors. Draw the network and identify the nodes and arcs. 

Part B: Solve the problem (provide exact values for all variables and the optimal objective function). 

#### Notes

Total Satisfaction table: 

| Professor | Fall Calc | Fall AI/ML | Fall Sim | Spring Calc | Spring AI/ML | Spring Sim |
| --- | --- | --- | --- | --- | --- | --- | 
| Prof 1 | 2 | 6 | 4 | 3 | 7 | 5 |
| Prof 2 | 6 | 5 | 8 | 4 | 3 | 5 |
| Prof 3 | 4 | 6 | 5 | 5 | 7 | 6 |

Arcs: 
- Source -> Professors: 3 arcs
- Professors -> course-semester nodes: 18 arcs
- Course-semester -> course nodes: 6 arcs
- Course nodes -> sink: 3 arcs
- 30 total arcs
- 12 nodes

Arc specifications: 

| Arc Layer | Cost | Lower Bound | Upper Bound|
| --- | --- | --- | --- | 
| Source -> Professor | 0 | 0 | 4 |
| Professor -> Course-semester | -satisfaction | 0 | 1 |
| Course-semester -> Course | 0 | 1 | 3 | 
| Course -> Sink | 0 | 0 | 4 |

```python
# === Define the Problem ===
prob = LpProblem("CourseAssignment", LpMinimize)

# Satisfaction Scores
satisfaction = {
    ('P1', 'Fall', 'Calc'): 2,
    ('P1', 'Fall', 'AIML'): 6,
    ('P1', 'Fall', 'Sim'): 4,
    ('P1', 'Spring', 'Calc'): 3,
    ('P1', 'Spring', 'AIML'): 7,
    ('P1', 'Spring', 'Sim'): 5,
    ('P2', 'Fall', 'Calc'): 6,
    ('P2', 'Fall', 'AIML'): 5,
    ('P2', 'Fall', 'Sim'): 8,
    ('P2', 'Spring', 'Calc'): 4,
    ('P2', 'Spring', 'AIML'): 3,
    ('P2', 'Spring', 'Sim'): 6,
    ('P3', 'Fall', 'Calc'): 4,
    ('P3', 'Fall', 'AIML'): 6,
    ('P3', 'Fall', 'Sim'): 5,
    ('P3', 'Spring', 'Calc'): 5,
    ('P3', 'Spring', 'AIML'): 7,
    ('P3', 'Spring', 'Sim'): 6,
}

# === Define Sets ===
professors = ['P1', 'P2', 'P3']
semesters = ['Fall', 'Spring']
courses = ['Calc', 'AIML', 'Sim']

# === Decision Variables ===
x = LpVariable.dicts("x", 
                     [(p, s, c) for p in professors
                      for s in semesters
                      for c in courses],
                     lowBound=0, upBound=1, cat='Integer')

# === Objective Function ===
prob += lpSum(-satisfaction[(p, s, c)] * x[(p, s, c)]
              for p in professors
              for s in semesters
              for c in courses)

# === Constraints ===
prob += lpSum(x[('P1', s, c)] for s in semesters for c in courses) == 4, "P1_load"
prob += lpSum(x[('P2', s, c)] for s in semesters for c in courses) == 4, "P2_load"
prob += lpSum(x[('P3', s, c)] for s in semesters for c in courses) == 4, "P3_load"
prob += lpSum(x[(p, s, 'Calc')] for p in professors for s in semesters) == 4, "Calc_sections"
prob += lpSum(x[(p, s, 'AIML')] for p in professors for s in semesters) == 4, "AIML_sections"
prob += lpSum(x[(p, s, 'Sim')] for p in professors for s in semesters) == 4, "Sim_sections"
prob += lpSum(x[(p, 'Fall', 'Calc')] for p in professors) >= 1, "Fall_Calc_min"
prob += lpSum(x[(p, 'Spring', 'Calc')] for p in professors) >= 1, "Spring_Calc_min"
prob += lpSum(x[(p, 'Fall', 'AIML')] for p in professors) >= 1, "Fall_AIML_min"
prob += lpSum(x[(p, 'Spring', 'AIML')] for p in professors) >= 1, "Spring_AIML_min"
prob += lpSum(x[(p, 'Fall', 'Sim')] for p in professors) >= 1, "Fall_Sim_min"
prob += lpSum(x[(p, 'Spring', 'Sim')] for p in professors) >= 1, "Spring_Sim_min"

# === Solve ===
prob.solve(PULP_CBC_CMD(msg=0))

# === Results ===
print(f"Status: {LpStatus[prob.status]}")
print(f"Maximum Satisfaction: {-value(prob.objective):.0f}")
print("\nAssignments:")
for p in professors:
    for s in semesters:
        for c in courses:
            if value(x[(p, s, c)]) == 1:
                print(f" {p} teaches {c} in {s} (satisfaction: {satisfaction[(p, s, c)]})")
```
