### Suppose the price of a BR increases from 50 to 60 and the price of a PC decreases from 80 to 50. Does the current solution remain optimal (in terms of the variables > 0 in the optimal solution)?

The allowable increase of 30 represents the maximum BR's coefficient can increase before the current basis changes. Since the reduced cost is exactly -30, BR is sitting right at the edge. Any increase beyond 30 would bring BR into the solution. 

100% Rule: 
- For each changing coefficient, calculate what fraction of its allowable range the change consumes. Then sum all those fractions. If the sum is $\le$ 1.0 (100%), the current basis is guaranteed to remain optimal
- BR: increases by 10, Allowable Increase = 30
- PC: decreases by 30, Allowable Decrease = 1E+30 (infinite, essentially)

$$\sum_{i} \frac{|\Delta c_i|}{\text{Allowable change in that direction}} \leq 1.0$$
$$\frac{10}{30} + \frac{30}{\infty} = 0.333 + 0 = 0.333 \le 1.0$$

The current basis remains optimal

```python
# BR increases -> use Allowable Increase
delta_BR = 60 - 50
allowable_increase_BR = 30
ratio_BR = delta_BR / allowable_increase_BR

# PC decreases -> use Allowable Decrease
delta_PC = 80 - 50
allowable_decrease_PC = 1e30
ratio_PC = delta_PC / allowable_decrease_PC

total_ratio = ratio_BR + ratio_PC
print(f"\nBR coefficient increases by {delta_BR}")
print(f" Ratio = {delta_BR} / {allowable_increase_BR} = {ratio_BR:.4f}")
print(f"\nPC coefficient decreases by {delta_PC}")
print(f" Ratio = {delta_PC} / {allowable_decrease_PC:.0e} = {ratio_PC:.6f} (= 0)")
print(f"\nSum of Ratios = {ratio_BR:.4f} + {ratio_PC:.6f} = {total_ratio:.4f}")
print()
if total_ratio <= 1.0:
    print("Sum <= 1.0 -> The current basis remains optimal.")
    print(" COLA = 33.33 stays in the solution.")
else:
    print(" Sum > 1.0 -> The current basis may no longer be optimal.")
```

### Suppose the price of a BR increases from 50 to 70 and the price of a PC increases from 80 to 90. Does the current solution remain optimal (in terms of the variables > 0 in the optimal solution)? 

The allowable increase of 30 represents the maximum BR's coefficient can increase before the current basis changes. Since the reduced cost is -30 and the proposed increase is 20, this falls in the allowable range. The allowable increase of 20 represents the maximum PC's coefficient can increase before the current basis changes. Since the reduced cost is -20 and the proposed increase is 10, this also falls in the allowable range. 

100% Rule: 
- BR: increases by 20, Allowable Increase = 30
- PC: increases by 10, Allowable Increase = 20

$$\sum_{i} \frac{|\Delta c_i|}{\text{Allowable change in that direction}} \leq 1.0$$
$$\frac{20}{30} + \frac{10}{20} = 1.1666667 > 0$$

The current basis is no longer optimal

```python
# BR increases -> use Allowable Increase
delta_BR2 = 70 - 50
allowable_increase_BR2 = 30
ratio_BR2 = delta_BR2 / allowable_increase_BR2

# PC increases -> use Allowable Increase
delta_PC2 = 90 - 80
allowable_increase_PC2 = 20
ratio_PC2 = delta_PC2 / allowable_increase_PC2

total_ratio2 = ratio_BR2 + ratio_PC2

print(f"\nBR coefficient increases by {delta_BR2}")
print(f" Ratio = {delta_BR2} / {allowable_increase_BR2} = {ratio_BR2:.4f}")
print(f"\nPC coefficient increases by {delta_PC2}")
print(f" Ratio = {delta_PC2} / {allowable_increase_PC2} = {ratio_PC2:.4f}")
print(f"\nSum of Ratios = {ratio_BR2:.4f} + {ratio_PC2:.4f} = {total_ratio2:.4f}")
print()
if total_ratio2 <= 1.0:
    print("Sum <= 1.0 -> The current basis remains optimal.")
    print(" COLA = 33.33 stays in the solution.")
else:
    print("Sum > 1.0 -> The current basis may no longer be optimal.")
    print(" A new variable may enter the basis.")
```

### How much would you pay for an additional unit of Resource 2 (Constraint 2)?

Resource 2: 
- RHS = 6, means we have 6 units available
- Final Value = 0, means we are using 0 units of it in the optimal solution
- Shadow Price = 0, represents the marginal value of one additional unit of a resource, means adding Resource 2 adds no value

Resource 2 is a non-binding constraint - we have 6 units available but use none of them. There is already slack in this constraint, so acquiring more would be wasteful. The shadow price of 0 confirms we should pay nothing for an additional unit. 

```python
shadow_price_R2 = 0
final_value_R2 = 0
rhs_R2 = 6

print(f"\nResource 2 Summary:")
print(f" Constraint RHS (current capacity): {rhs_R2}")
print(f" Final Value (amount used): {final_value_R2}")
print(f" Shadow Price: {shadow_price_R2}")
print()
print("Interpretation:")
print(f" Since Final Value ({final_value_R2}) < RHS ({rhs_R2}), Resource 2 is NOT fully used.")
print(f" It is a non-binding constraint (slack > 0).")
print(f" A Shadow Price of {shadow_price_R2} means an additional unit of Resource 2")
print(f" provides zero additional value to the objective.")
print()
print(f" Answer: We would pay 0 for an additional unit of Resource 2.")
print(f" There is already unused capacity - acquiring more would be wasteful.")
```

### Suppose you were to acquire 150 additional units of Resource 1 (Constraint 1), what would your new objective value be?

If the change exceeds the allowable range, the basis changes, meaning the set of active constraints shifts. Once that happens, the shadow price of 2 no longer applies and we cannot use it to calculate the new objective. We would need to re-solve the entire LP

Shadow Price is like a local rate - it's only valid within a certain neighborhood of the current solution. 

- Proposed change = +150 units of Resource 1
- Allowable Increase = 1E+30 (infinite)
- $150 \le \infty$

Final calculation: 
$$\text{New Objective} = \text{Current Objective} + (\text{Shadow Price} * \Delta RHS)$$
$$\text{New Objective} = 1000 + (2*150)$$
$$\text{New Objective} = 1300$$

Every additional unit of Resource 1 is worth 2 to us. By acquiring 150 more units, we need to improve our objective by 300, bringing us from 1000 to 1300

```python
shadow_price_R1 = 2
current_rhs_R1 = 500
allowable_increase_R1 = 1e30
allowable_decrease_R1 = 380
delta_rhs = 150
current_objective = 1000

print(f"\nResource 1 Summary:")
print(f" Current RHS: {current_rhs_R1}")
print(f" Shadow Price: {shadow_price_R1}")
print(f" Allowable Increase: Infinite (1E+30)")
print(f" Allowable Decrease: {allowable_decrease_R1}")
print(f" Proposed delta RHS: +{delta_rhs}")
print()

# Check within range
within_range = delta_rhs <= allowable_increase_R1
print(f"Is +{delta_rhs} <= Allowable Increase ({allowable_increase_R1:.0e})? {'Yes' if within_range else 'No'}")
print()

if within_range: 
    new_objective = current_objective + shadow_price_R1 * delta_rhs
    print(f"New Objective = Current Objective + (Shadow Price x delta RHS)")
    print(f" = {current_objective} + ({shadow_price_R1} x {delta_rhs})")
    print(f" = {current_objective} + {shadow_price_R1 * delta_rhs}")
    print(f"\n New Objective Function Value = {new_objective}")
```
